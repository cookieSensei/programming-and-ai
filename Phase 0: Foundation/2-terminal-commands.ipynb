# Navigating the Terminal

Before we run our Python script, let's learn a few basic terminal commands.

You don't need to become a Linux expert.

We only want to learn enough to **move around our project files and run our programs**.

Imagine your computer's files as a set of folders:

```text
home/
└── student/
    ├── projects/
    │   ├── app.py
    │   └── notes.txt
    └── downloads/
```

The terminal lets us move through this structure.

---

## Where am I?

Use:

```bash
pwd
```

`pwd` means **Print Working Directory**.

It tells you which folder you are currently inside.

For example:

```text
/home/student/projects
```

Think of it as asking:

> **"Where am I right now?"**

---

## What is inside this folder?

Use:

```bash
ls
```

`ls` shows the files and folders in your current location.

For example:

```text
app.py
notes.txt
projects
```

Think of it as:

> **"What is around me?"**

---

## Move into a folder

Use:

```bash
cd folder
```

`cd` means **Change Directory**.

For example, if you have:

```text
projects/
    app.py
    python/
```

you can move into the `python` folder:

```bash
cd python
```

Your current location has now changed.

You can check it with:

```bash
pwd
```

And see what's inside with:

```bash
ls
```

---

## Move back one folder

Use:

```bash
cd ..
```

The `..` means:

> **"The folder above my current folder."**

For example:

```text
projects/
└── python/
    └── app.py
```

If you are inside `python`:

```bash
cd ..
```

you will move back to:

```text
projects/
```

You can think of it like going **up one level** in the folder structure.

---

## Move into a folder and back out

Let's practice.

Start by looking at where you are:

```bash
pwd
```

See what's there:

```bash
ls
```

Find a folder and move into it:

```bash
cd folder
```

Look around:

```bash
ls
```

Then go back:

```bash
cd ..
```

Check your location again:

```bash
pwd
```

You have now navigated through the filesystem using the terminal.

---

## The four commands to remember

For now, these four commands are enough:

| Command     | What it does        |
| ----------- | ------------------- |
| `pwd`       | Shows where you are |
| `ls`        | Shows what's here   |
| `cd folder` | Moves into a folder |
| `cd ..`     | Moves up one folder |

A useful mental model is:

```text
pwd  → Where am I?
ls   → What's here?
cd   → Move somewhere
..   → Go up one level
```

---

## Finding `app.py`

Let's connect this back to our Python program.

Suppose your project looks like this:

```text
projects/
└── my-first-program/
    └── app.py
```

You can navigate to your project with:

```bash
cd projects
```

Then:

```bash
cd my-first-program
```

Check what's inside:

```bash
ls
```

You should see:

```text
app.py
```

Now you can run your program:

```bash
python app.py
```

Notice the workflow:

```text
Navigate to project
        ↓
Find app.py
        ↓
Run app.py
        ↓
See the output
```

This is why we're learning the terminal.

We're not memorizing random commands.

We're learning how developers **move around their projects and run their code**.

---

## One More Useful Command

You can clear the terminal with:

```bash
clear
```

This doesn't delete anything.

It simply clears the text currently visible in your terminal.
