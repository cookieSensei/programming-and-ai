```python
some_str = "this is A random strinG"
```

```python
re.findall("a", some_str)
```

re.findall iterates over a string, character by character and searches for a given pattern. So, in this case. In the first iteration of re.findall("a", some_str), re will try to match the pattern "a" with the first character of our string, that is "t"

The string "this is A random strinG" starts from "t" right...

first iteration: pattern is matched with "t", pattern is "a", no result, pattern didn't get matched

Second iteration: pattern is match with "h" and so on till the length of the whole string till all "a" are extracted out


```python
pattern = 'r'

re.findall(pattern, some_str)

# Extracted are all the character set matched to the pattern
```

```python
re.findall("this", some_str)
```

```python
re.findall("thiS", some_str) # thiS does not exist
# this do exist but
# this and thiS are different as python is case sensitive
```

```python
"this" == "thiS"
```

```python
"this" == "this"
```

```python
# In every iteration, check if the text string is a
# character pattern inside []
some_str = "G this is A random strinG"

re.findall("[Gabcg]", some_str)
```

```python
# In every iteration, check if the character is either a or r

some_str = "this is A random strinG"

re.findall("[ar]", some_str)
```

```python
# In every iteration, check if the character is either a, b or r

re.findall("[abr]", some_str)
```

In every iteration, check if the character is either a, b or t

```python
some_integers = "1234 431 776 2341"

pattern = "[23][41]" # feched out string should be of length 2
# first character should be either 2 or 3,
# while second character should be either 4 or 1

re.findall(pattern, some_integers)
```

```python
# I'll change the string a little
some_integers = "123333224 43321 7372326 2341"

# + means that the pattern will be fetched out one or more time
pattern = "[23]+"

# So long as the pattern keeps matching, the pattern will be part of the
# same string in the output
re.findall(pattern, some_integers)
```

```python
# I'll change the string a little
some_integers = "123333224 43321 7372326 2341"

# * means that the pattern will be fetched out zero or more time
# if there's no pattern in the string, empty string is returned
pattern = "[23]*"

# So long as the pattern keeps matching, the pattern will be part of the
# same string in the output
re.findall(pattern, some_integers)
```

```python
some_integers = "123333224 43321 7372326 2341"

pattern = "[23]3"

re.findall(pattern, some_integers)
```

Notice how the second character of the returned string is always 3
while the first character could take either 2 or 3
So, only 2 possible outputs: 33 or 23


```python
some_integers = "123333224 43321 7372326 2341"

pattern = "[23][3]+"
# is not same as pattern: "[23]+"

# pattern "[23][3]+" means match 2 or 3 at first match and
# after that keep match any 3 present in the string

re.findall(pattern, some_integers)
```

```python
some_integers = "123333224 43321 7372326 2341"

pattern = "[23]+"

re.findall(pattern, some_integers)
```

## Greedy matching


If we have two character in a series of pattern matches, say the symbol _ appears two or more times. Where to end the pattern match now, at the first encounter of underscore or the last encounter of underscore character _


```python
re.findall(".+", "this\nis")
```

```python
# re tends to get greedy by default

import re
x = 'Lets _ use the _ underscore character'
y = re.findall('.+_', x)
print(y)
```

re extracted the _ character from the end


```python
import re
x = 'Lets _ use the _ underscore character'
y = re.findall('.+?_', x)
print(y)
```

```python
re.findall("\s[\w\.]+@[\.\w]+.com", x)
```

### HTML tags


```python
sample_html = """
<body>
<h1>Heading 1</h1>
<p>This is some random paragraph.
<br>I have forgotten a lot about how to write html.</p>
<p>But that's not the objective. Objective is to get specific tags.
Anyhow, if I had to work in HTML I'd just google things that I may have forgotten.
</p>
</body>
"""

print(sample_html)
```

If we search for the html tag pattern in our usual way, we'll encounter the greedy behaviour of re


```python
re.findall("<.+>", sample_html)
```

In the above multi line string we said, match the character < and then match any non line character till we encounter the character >

Our objective is also to find these < tags > but we need some way to tell re, wait stop at the first > character you encounter.


```python
re.findall("<.+?>", sample_html)
```

# re.search()


re.search() returns a yes (True) or no (False) depending on if the search argument was found or not


```python
if re.search("a", "sample string"):
    print("This block was executed")
else:
    print("pattern not found")
```

```python
if re.search("at", "sample string"):
    print("This block was executed")
else:
    print("pattern not found")
```
