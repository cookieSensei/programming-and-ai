# Retrieving Information from the Web with BeautifulSoup

So far, our programs have mostly worked with information that we write ourselves.

Now we're going to make our Python program **read information from a website**.

For this, we'll use two libraries:

* `requests` - to download a webpage
* `BeautifulSoup` - to understand and extract information from the HTML

Our goal is simple:

> **Give Python a URL → download the webpage → extract useful information from it.**




# 2. Download a Webpage

Let's start by downloading a webpage.

```python
import requests

url = "https://cookiesensei.com"

response = requests.get(url)

print(response)
```

You should see something similar to:

```text
<Response [200]>
```

The number `200` is an HTTP status code.

It means:

> **The request was successful.**

We can also look at the actual webpage that was downloaded:

```python
print(response.text)
```

This will produce a lot of text.

You might see things such as:

```html
<html>
<head>
    <title>CookieSensei</title>
</head>

<body>
    <h1>Learn by Building</h1>

    <p>Build software. Learn by doing.</p>
</body>
</html>
```

This is **HTML**.

---

````python
url = "https://cookiesensei.com" 
response = requests.get(url) 
print(response)
````


# 3. What is HTML?

HTML is the language used to structure webpages.

For example:

```html
<h1>Hello</h1>
```

means that `Hello` is a heading.

And:

```html
<p>Hello world!</p>
```

means that `Hello world!` is a paragraph.

A webpage is essentially a tree of HTML elements.

For example:

```text
html
├── head
│   └── title
│
└── body
    ├── h1
    ├── p
    └── p
```

When we download a webpage with `requests`, Python receives this HTML.

Now we need a way to work with it.

That's where **BeautifulSoup** comes in.

---

# 4. Create a BeautifulSoup Object

Let's give the HTML to BeautifulSoup:

```python
import requests
from bs4 import BeautifulSoup

url = "https://cookiesensei.com"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")
```

Now `soup` represents the webpage in a form that Python can search and explore.

Think of it like this:

```text
Website
   ↓
requests
   ↓
HTML
   ↓
BeautifulSoup
   ↓
Python-friendly webpage structure
```

---

````python
soup = BeautifulSoup(response.text, "html.parser")
print(soup)
````


# 5. Find the Page Title

Let's retrieve the title of the webpage.

```python
title = soup.title

print(title)
```

You might get:

```html
<title>CookieSensei</title>
```

If we only want the text inside the title:

```python
print(soup.title.text)
```

Now we get:

```text
CookieSensei
```

---

````python
title = soup.title
print(title)
````


# 6. Find a Heading

We can search for HTML elements by their tag.

For example:

```python
heading = soup.find("h1")

print(heading)
```

If the webpage contains:

```html
<h1>Learn by Building</h1>
```

we'll get:

```html
<h1>Learn by Building</h1>
```

To get only the text:

```python
print(heading.text)
```

Output:

```text
Learn by Building
```

---

````python
heading = soup.find("h1")
print(heading)

````

````python
print(heading.text)
````


# 7. Find a Paragraph

We can do the same thing with paragraphs:

```python
paragraph = soup.find("p")

print(paragraph.text)
```

This gives us the text inside the first `<p>` element.

---

````python
paragraph = soup.find("p")
print(paragraph.text)
````


# 8. Find All Elements

Sometimes we don't want just the first paragraph.

We want **all paragraphs**.

For this, use:

```python
paragraphs = soup.find_all("p")
```

Now `paragraphs` contains all the paragraph elements.

We can loop through them:

```python
for paragraph in paragraphs:
    print(paragraph.text)
```

This is a very common pattern when scraping webpages:

```text
Find all elements
       ↓
Loop through them
       ↓
Extract information
```

---


````python
paragraphs = soup.find_all("p")

for paragraph in paragraphs:
    print(paragraph.text)
````



# 9. Extract All Links

Links are represented by the `<a>` HTML tag.

For example:

```html
<a href="/about">About</a>
```

We can find all links:

```python
links = soup.find_all("a")
```

Then:

```python
for link in links:
    print(link.text)
```

This prints the text displayed for each link.

But often we want the actual URL.

The URL is stored inside the `href` attribute:

```python
for link in links:
    print(link.get("href"))
```

For example:

```text
/about
/curriculum
/contact
```

---

````python
links = soup.find_all("a")

for link in links:
    print(link.text)
````




# 10. Getting Both Link Text and URL

We can retrieve both:

```python
for link in links:

    text = link.text
    url = link.get("href")

    print(text, "→", url)
```

We might get:

```text
About → /about
Curriculum → /curriculum
Contact → /contact
```

This is exactly the kind of information our web crawler will need.

Our crawler can:

1. Download a webpage.
2. Find all the links.
3. Follow those links.
4. Download those pages.
5. Extract their text.

---

````python
for link in links:
    text = link.text
    url = link.get("href")
    print(text, ">", url)
````


# 11. Extract All the Text

Sometimes we don't care about individual HTML elements.

We simply want the readable text from the webpage.

BeautifulSoup gives us:

```python
text = soup.get_text()
```

We can make the output cleaner:

```python
text = soup.get_text(" ", strip=True)

print(text)
```

Now we get something closer to:

```text
CookieSensei Learn by Building ...
```

rather than the HTML itself.

This is useful for our chatbot because **the text of the webpage will become its memory**.

---

````python
text = soup.get_text(" ", strip=True)
print(text)
````


# 12. Removing Unwanted Elements

A webpage contains much more than the information we want.

It may contain:

* JavaScript
* CSS
* navigation
* footers
* menus
* tracking code

Before extracting the text, we can remove some of these elements.

For example:

```python
for element in soup(["script", "style"]):
    element.decompose()
```

Then:

```python
text = soup.get_text(" ", strip=True)
```

Now we're much closer to the actual readable content of the page.

---

````python
for element in soup(["script", "style"]):
    element.decompose()

text = soup.get_text(" ", strip=True)
````

# 13. Putting It All Together

Here's our complete small example:

```python
import requests
from bs4 import BeautifulSoup


url = "https://cookiesensei.com"

response = requests.get(url)

soup = BeautifulSoup(
    response.text,
    "html.parser"
)


# Remove things we don't need
for element in soup(["script", "style"]):
    element.decompose()


# Get the page title
print("TITLE:")
print(soup.title.text)


# Get all headings
print("\nHEADINGS:")

for heading in soup.find_all(["h1", "h2", "h3"]):
    print(heading.get_text(" ", strip=True))


# Get all links
print("\nLINKS:")

for link in soup.find_all("a"):
    text = link.get_text(" ", strip=True)
    url = link.get("href")

    print(text, "→", url)


# Get all readable text
print("\nPAGE TEXT:")

text = soup.get_text(" ", strip=True)

print(text)
```

---

````python
import requests 
from bs4 import BeautifulSoup 

url = "https://cookiesensei.com" 

response = requests.get(url) 

soup = BeautifulSoup( response.text, "html.parser" ) # Remove things we don't need 

for element in soup(["script", "style"]): 
    element.decompose() 
    
# Get the page title 
print("TITLE:") 
print(soup.title.text) 

# Get all headings 
print("\nHEADINGS:") 

for heading in soup.find_all(["h1", "h2", "h3"]): 
    print(heading.get_text(" ", strip=True)) 
    
    
# Get all links 
print("\nLINKS:") 

for link in soup.find_all("a"): 
    text = link.get_text(" ", strip=True) 
    url = link.get("href") 
    
    print(text, "→", url) 
    
# Get all readable text 

print("\nPAGE TEXT:") 

text = soup.get_text(" ", strip=True) 
print(text)
````
