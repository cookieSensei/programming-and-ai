# For loop

Now we are ready to learn a simple for loop.

Lets imagine if our object was to print a same thing 5 times. We would have to in this case write print function 5 times, right?

```python
4 in range(5)
```

```python
for i in range(5):
    print("Hello. THe value of i is: {}".format(i))
```

"**Tickles** what was that sorcery"

This is the ability of loops to repeat tasks.

To understand this, let's look at a simple example first.

Our problem statement is to print individual elements inside a list one by one.

```python
list1 = [10,20,30,40]
list1
```

We could use the print function 4 times here keeping in mind the appropriate indexes for the list elements

```python
print(list1[0])
print(list1[1])
print(list1[2])
print(list1[3])
```

But if the list contained say 20 elements, would you still be willing to print all elements one by one? No right.

In that case, we start a for loop with.

**for i in list1**

```python
list1 = [1,2,4,7]
```

```python
for i in list1:
    print(i)
```

When python sees the keyword for, python starts a for loop, then python sees this "in" keyword.

Remember the in keyword?

```python
10 in list1
```

```python
for i in [1, 2, 3]:
    print(i)
```

With this code, python started a loop, initiated

```python
# When we write 

# for i in range(5):

# Python starts a loop (for loop) 
```

```python
for i in range(5): # This will produce 5 True boolean
    print(i) # i will be printed 5 times
```
