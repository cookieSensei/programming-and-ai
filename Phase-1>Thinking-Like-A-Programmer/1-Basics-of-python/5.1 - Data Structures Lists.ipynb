## Subsetting / Indexing elements of a sequence

You might have figured out by now that list is a container of multiple elements and so I like to call it a sequence data type.
We will talk about this topic on detail when we reach **Numpy** array. For now, just have the intuition that we can extract individual elements from sequence data types.

```python
a1 = 10
a2 = 10.1
a3 = "Something"
a4 = True
a5 = False
a6 = None

list1 = [a1, a2, a3, a4, a5, a6, 700]
print(list1)
```

```python
len(list1)
```

New ability unlocked: len() function
Uses: calculate the length of a sequence

Now, let's learn some ways of accessing individual elements inside a sequence data type.


```python
print(list1[0])  # This will give us the first element of our sequence
```

Indexing is done using the square brackets. Indexing in python start from 0.

```python
print(list1[1])   # This will give us second element
```

If we want to access the last element of a sequence, we can use negative indexing of Python.

```python
print(list1[-1])  # This will give us the last element
```

```python
print(list1[-2])  # Second last element
```

If our objective is to extract a range of elements, we could use the : operator.


```python

# list_name[first_index (included): second_index (excluded)]
```

```python
print(list1[0:2])  # The first index is included, 
                   # while the second index is excluded
    # index 0 and index 1 elements will be returned
```

```python
print(list1)
```

```python
print(list1[0:3])
```

```python
print(list1[0:4])
```

```python
print(list1)
```

```python
print(list1[1:3])   # second element (index 1) included, 
                    # fourth element (index 3) excluded
```

```python
print(list1[0:]) # print everything starting from index 0
```

```python
print(list1[1:]) # print everything starting from index 1
```

```python
print(list1[3:]) # print everything starting from index 3
```

```python

# list_name[first_index (included): second_index (excluded) : step]
```

```python
list1
```

```python
print(list1[0:3:2])  # start from index 0, go till index 2 in the steps of 2
```

```python
list1
```

```python
# print list starting from index 0 till index 3 with the step of 1
print(list1[0: 4: 1])
```

```python
# print list starting from index 0 till index 3 with the step of 2
print(list1[0: 4: 2])
```

### List methods 

There are some useful in-built functions that are available for the list data type. Let's look at some of them. You are not expected to cram any of these functions. 

```python
list2
```

**append** method

Just add a new element at the end of the list 

```python
list2 = list(range(10))
list2.append("new element")
```

```python
list2
```

**pop** method

Remove an element from the list and return it to the user. Its actual working will be understood after the function chapter. So, till then, just stick with the easy explaination, that it removes an element from a list.

```python
list2.pop(-1) # Remove the last element from the list2
```

```python
list2.pop(0) # Remove the first element from the list2
```

```python
list2
```

We can make changes to the elements inside a list. Let's say we want to change the value of first element of list to 100.

```python
list2 = list(range(10))
list2
```

```python
list2[0]
```

```python
list2[0] = 100 # Just use proper indexing to reference the element inside a list and then assign a new value
```

```python
list2
```

Let's say we want to add 1000 to the last element.
First, reference the last element and then add 1000 to it and assign this new value to the old reference (appropriate index)

```python
list2[-1] = list2[-1] + 1000
```
