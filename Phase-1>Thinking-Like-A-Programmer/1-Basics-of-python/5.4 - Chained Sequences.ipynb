# Lists within a list

So far, we have discussed lists, tuples, dictionaries, sets as a sequence.

But we can chain these data types together to create say, a list within a list, list within a tuple and so on.

```python
list1 = list(range(4))
list1
```

```python
list2 = ["one", "two", "three", "four"]
list2
```

We have the append method to append anything inside a list. Lets see what happens when we append a list to another list.

```python
list1.append(list2)
```

```python
list1
```

```python
list1[0]
```

```python
list1[-1]
```

As we can see, the last element of the list1 is the list2 itself. We may or may not want this behaviour.

What if our objective was to add just the elements together

```python
list3 = list(range(4))
list3
```

```python
list4 = ["one", "two", "three", "four"]
list4
```

```python
list3.extend(list4)
```

```python
list3
```

In this case, notice that individual elements from the list4 are added inside the list3.

```python
list3[-1]
```

```python
list1[-1]
```

```python
list1
```

```python
list3
```

# Tuple within a list

Lets create a tuple within a list

```python
list5 = [1, 2, 3, (7,8,9)]
```

```python
list5
```

```python
list5[-1]
```

Notice that the last element of list is a tuple.

We remember that once a tuple is created, we cannot change the tuple elements. Although, we can reference that value.

```python
list5[-1][0]
```

```python
list5[-1][1]
```

```python
list5[-1][2]
```

# Dictionary inside a list

```python
list6 = [1, 2, 3, {"Name": "Tickles", "Age": 29}]
list6
```

```python
list6[-1]
```

Last element inside the list is a dictionary, and we index dictionaries by keys

```python
list6[-1]["Name"]
```

```python
list6[-1]["Age"]
```

So on, we can create lists inside tuples, and vice versa. We can create dictionaries inside tuples or lists, or we could create lists inside a dictionary.

This is one of the most important topic to understand, but it comes with time and practice. When we reach the module numpy and pandas, you'll see how useful chaining different data types are.
