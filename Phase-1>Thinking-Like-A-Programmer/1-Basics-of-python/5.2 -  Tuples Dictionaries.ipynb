# Tuples

Tuples is a sequence data type as well. We can keep different elements inside tuples. Tuples are a combination of different elements separated by a comma.

Tuples are also like a list but with a change that we can not change indivudual values inside a tuple. We can reference those individual elements though.

That makes a tuple immutable. Lists are mutable. Mutable means, we can change the individual elements.


```python
tuple1 = 1, 2, "third element", True
```

```python
tuple1
```

You might have noticed that list uses square brackets to store elements. Tuples uses round brackets to store elements. Just don't be confused as of now with these bracket notations.

Just stick to the notation that list uses square brackets and tuples use round brackets.

```python
tuple2 = (1, 2, "third element", True)
```

```python
tuple2
```

```python
tuple1[0]   # Get the first index of tuple1
```

```python
tuple1[:-1]   # get all elements except the last index
```

```python
# This cell will throw an error as we will attempt
# to change an individual value here

# tuple1[0] = "New element"
```

We can change a list to a tuple by using the tuple() function

```python
list3 = list(range(10))
list3
```

```python
tuple2 = tuple(list3)
tuple2
```

# Dictionary

Dictionary stores different data types as a key value pair.

$$Key : Value$$

```python
dict1 = dict(Name = "Tickles",
             Age = 29,
             Occupation = "Jobless",
             Excuse = "I'm a lead AI engineer for Python"
            )
dict1
```

We usually don't use this way to create dictionaries as this confuses a lot of people as to why the syntax for creating dictionaries is so different.

If a character sequence in python has a string data type, then why Name, Age, Occupation, Excuse are not typed inside the quotes.

Well, a lot of these things will be cleared by the time we've reached the **function** chapter.

For now, let's look at an easier way which we all use to create dictionaries.

```python
dict2 = {
    "Name" : "Tickles",
    "Age" : 29,
    "Occupation" : "Jobless",
    "Excuse" : "I'm a lead AI engineer for Python"
}
dict2 
```

```python
dict2["Name"]
```

Now notice, no matter the way we chose to create the dictionary, notice in the output, that dictionaries have the curly brackets / braces and that the key value pairs are displayed with this notation 

$$Key:Value$$

Now let's look at some useful operations with dictionaries to clear why this particular data type is so important.

```python
dict2
```

```python
dict2["Name"]
```

### Indexing with dictionaries

Individual values are stored inside a dictionaries as **Values** and we use some unique **Keys** to reference them.

$$Dictionary[Key] = Value$$

As you might have noticed, indexing with dictionaries is different from indexing in lists and tuples.

```python
dict2["Age"]
```

We can overwrite the indivudual values stored inside a dictionary. But let's first learn how to add values to a dictionary.

```python
dict2["Salary"] = "I do get paid. No trust me I earn. Alright you got me"
```

```python
dict2
```

```python
dict2["Magic power"] = "1000 magic points"
```

```python
dict2
```

Now let's look at how we can delete individual elements of a dictionary.

Lets delete the value of the key "Salary" as **Tickles** clearly got offended with this question.

We will use the del keyword here.


```python
del dict2["Salary"]
```

```python
dict2
```
