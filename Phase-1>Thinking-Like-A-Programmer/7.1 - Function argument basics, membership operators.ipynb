# Range() function

You could recall that we used a range() function to generate lists. Actually, range() is just a sequence generator. range() is a python object (which we will cover in depth later)

So, if we make a range(10) and print its output, it'd look something like this.

```python
range(10)
```

Its when we create a sequence from it that it generates those numbers.

For now, we will generate numbers using a list function. So, those numbers will be created inside a list.

```python
list(range(10))
```

```python
list(range(4))
```

Depending on the value we feed inside the range function, we get numbers generated from zero till that value. 

$$list(range(4))$$ gives us 4 numbers starting from zero.

$$[0,1,2,3]$$

range accepts a whole range of how many numbers one wants, where to start the sequence from, where to end the sequence, and many more operations.

```python
list(range(0, 5))
```

```python
# list(range(start_point, end_point))
```

```python
list(range(2, 5))
```

As you might have noticed, the end point value is excluded from the range.

$$list(range(2, 5))$$

gives us a range of numbers starting from 2 ending at 4 as the end point is excluded.

```python
list(range(10, 20))
```

These numbers inside range(10, 20) separated by a comma that we enter inside the range functions are called **function arguments**.

# Membership Operators

Remember, we left the membership operators for later.

There are two membership operators in python. 

They are used to check if a value exists inside a sequence.

### The in keyword

```python
1 in [1, 2, 3]
```

Python says, yes the value 1 exists inside the list of elements [1, 2, 3]

```python
value1 = 10
value1
```

```python
sequence_list = [1, 2, 3, 10]
sequence_list
```

```python
value1 in sequence_list
```

```python
12 in sequence_list
```

Python says the value 12 does not exist inside the list.

### The not in operator

Just the opposite of in operator, because we used not keyword here to reverse the output

```python
sequence_list
```

```python
1 in sequence_list
```

```python
1 not in sequence_list
```
