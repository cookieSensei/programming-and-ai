# Assignment Operators

### The = operator

```python
# The = operator assigns some data to a variable

c = 12
print(c)
```

Let's increase the value of a variable to understand the next assignment operators

```python
# Counter intuitive to how we use = sign in our mathematics class
# = sign in python just assigns the right hand side RHS of = sign
# to the left hand side LHS of = sign

d = 5

# we will do d+1 first
# and then assign the increased value to the same variable d

d = d + 1   
print(d)
```

### The += operator

To perform the operation d=d+1 in a pythonic way, we use += operator. Nothing fancy, even I don't use it that often, but other people do use it. So, it becomes important to understand such operators.

```python
# Initialize the value 9 as variable e

e = 9

# Increase the value of variable e by one
e += 1

print(e)
```

```python
f = 10

f += 15   # Again, the same thing as f=f+15

print(f)
```

### The -= operator



```python
g = 7

g -= 2   # Pythonic way of: g=g-2

print(g)
```

### The *= operator

```python
h = 5

h *= 2    # Pythonic way of: h=h*2

print(h)
```

### The /= operator

```python
i = 6

i /= 2    # Pythonic way of: i=i/2

print(i)
```

### The %= operator

```python
j = 10

j %= 3    # Pythonic way of: j=j%3, remainder

print(j)
```

### The //= operator

### The **= operator

```python
l = 10

l **= 3    # Pythonic way of: l=l**3

print(l)
```

# Comparison Operators

### The == operator

Remember the data type boolean, this is where boolean becomes extremely useful.

The == operator compares the RHS right hand side of the  == sign to the LHS left hand side of the == sign.

The output is hence returned in True or False depending on if the condition was true or false.

```python
# Let's compare the integer 4 to the integer 5

# Here we will ask python if the value 4 equals the value 5
# Of course it isn't, that's why python will say False

4 == 5
```

```python
4 == 4
```

### The != operator

This is the not equal operator. It will do something opposite of what == does. It asks python if the LHS is not equal to the RHS.

Yes, 4 does not equal 5, that's why python says yes the condition is true and python returns True.

# Identity Operators

### The is operator

Unlike the == operator which checks if two objects of python have same value, the "is" operator checks if two objects are one and the same.

**Note**: Please pay close attention now

```python
x = 4
y = 4

x is y
```

At this point you might be willing to think that of course x is same as y as both have the value 4. Lets make a more complicated example.


```python
x = "random word"

y = "random word"

x is y
```



When our data type is very simple like an integer, python don't create a new object in memory (RAM) for the different objects having same value.

So, when we create the data integer 4 and assign it to variable x, python creates the integer 4 and names it x

```python
x = 4
```

Now, if we create another integer 4 and call it y, python will use the integer value 4 that is already stored in python and would give it another name y.

```python
y = 4
```

Here, instead of creating a new integer 4, python will re-use the integer 4 from memory and give it another name y. So now, that integer 4 has two names, x and y.

```python
x is y
```

Python says, x and y are two same objects in memory. Let's look at their memory locations.


#### id() function


id() function returns the memory location of objects inside python. In simple words, it confirms for us if different variable names are given to just one single object of python.

```python
id(x)
```

```python
id(y)
```

Things get tricky when our data type is mute-able (We'll talk more about it later)

Let's say we have a sequence data type like a string.

Now, if we create two strings of same value, but use different variable names to reference them, python will give it two different memory locations, so these two string variables now become two different objects in memory.

```python
string1 = "a "
string2 = "a "

string1 is string2
```

These two sequence data type might have same value but they are stored in different memory locations inside python. So, in this case == operator will say they have same value. The is operator will say they have different memory locations, so they are not the same object.

```python
string1 == string2
```
