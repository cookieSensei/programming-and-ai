# Classes

Classes can be understood as a data type. It is not a data type, but for intuition can be assumed as a custom data type.

Like for example, here we will create a custom data type Animal which will have features just like an animal does.

```python
class Animal:
    """A class representing an animal with a strength attribute."""

    strength = 10  # Default strength value for all animals

    def strength_plus(self, strength_points):
        """
        Increases the animal's strength by a given amount.

        Parameters:
        strength_points (int or float): The amount to increase strength by.
        """
        self.strength = self.strength + strength_points  # Update the strength attribute
```

```python
animal1 = Animal()

animal1.strength_plus(1000)

animal1.strength
```

```python
panda = Animal() # creating an instance of the Animal class
```

```python
print(f"Initial strength of panda: {panda.strength}")  # Output: 10 (default strength)

panda.strength_plus(5)  # Increases strength by 5
print(f"Strength of panda after increase: {panda.strength}")  # Output: 15
```

```python
# Define a class named 'Species'
class Species():

    # The __init__ method is the constructor that initializes the object
    # It takes type_of_species, name, and power as arguments when creating an instance of the class
    def __init__(self, type_of_species, name, power):
        # Initialize instance variables with the provided values
        self.type_of_species = type_of_species  # Set the 'type_of_species' (e.g., animal type)
        self.name = name  # Set the 'name' of the species
        self.power = power  # Set the 'power' attribute of the species
    
    # Define a class variable 'strength' with a default value of 10
    strength = 10

    # Method to increase the 'strength' by a given value 'strength_points'
    def strength_plus(self, strength_points):
        # Add the value of 'strength_points' to the 'strength' attribute
        self.strength = self.strength + strength_points
    
    # The __add__ method allows the use of the + operator to add two 'Species' objects
    # It returns the sum of their 'power' attributes
    def __add__(self, other):
        # 'self' refers to the current instance, and 'other' refers to the second instance
        return self.power + other.power
```

```python
# Create an instance of the Species class
# The arguments "Panda", "Bill", and 50 are passed to the constructor (__init__) to initialize the object

var1 = Species("Panda", "Bill", 50)
```

```python
# Access the 'name' attribute of the var1 object
print(var1.name)  # Output: "Bill"
```

```python
# Access the 'type_of_species' attribute of the var1 object
print(var1.type_of_species)  # Output: "Panda"
```

```python
# Access the 'power' attribute of the var1 object
print(var1.power)  # Output: 50
```

```python
# Access the 'strength' attribute of the var1 object
print(var1.strength)  # Output: 10
```

```python
# Decrease the strength by 10
var1.strength_plus(-10)

# Access the 'strength' attribute of var1 after the update
print(var1.strength)  # Output: 0
```

```python
# Create a second instance of the Species class with type "Cat", name "FoxyPirates", and power 5
var2 = Species("Cat", "FoxyPirates", 5)
```

```python
var1+var2
```

```python
# Accessing the 'type_of_species', 'name', and 'power' attributes of var2
print(f"Species Type: {var2.type_of_species}")  # Output: "Cat"
print(f"Species Name: {var2.name}")  # Output: "FoxyPirates"
print(f"Species Power: {var2.power}")  # Output: 5

# Accessing the default strength attribute of var2
print(f"Species Strength: {var2.strength}")  # Output: 10
```

```python
# Define a class named 'Butterfly'
class Butterfly():
    
    # The __init__ method is the constructor that initializes the object
    # It takes name, color, and season as arguments. If season is not provided, it defaults to 'Autumn'.
    def __init__(self, name, color, season='Autumn'):
        self.name = name  # Set the 'name' of the butterfly
        self.color = color  # Set the 'color' of the butterfly
        self.season = season  # Set the 'season' of the butterfly (default is 'Autumn')
    
    # Define a class variable 'beauty_points' with a default value of 100
    beauty_points = 100
    
    # Method to increase or decrease the 'beauty_points' by a given value 'points'
    # The 'increase' argument controls whether to add or subtract points
    def beauty_plus(self, points, increase=True):
        if increase:
            self.beauty_points += points  # Increase beauty_points by 'points' if 'increase' is True
        else:
            self.beauty_points -= points  # Decrease beauty_points by 'points' if 'increase' is False
            
    # The __add__ method allows the use of the + operator to add two 'Butterfly' objects
    # It returns the sum of their 'beauty_points' attributes
    def __add__(self, other):
        # 'self' refers to the current instance, and 'other' refers to the second instance
        return self.beauty_points + other.beauty_points
```

```python
# Create an instance of Butterfly with name 'Queen Bee' and color 'Golden Blue'
queenB = Butterfly('Queen Bee', 'Golden Blue')
```

```python
queenB
```

```python
# Accessing the 'season' attribute of queenB
print(queenB.season)  # Output: "Autumn"
```

```python
# Subtracting 20 from beauty_points
queenB.beauty_plus(20, False)

# Accessing the updated beauty_points
print(queenB.beauty_points)  # Output: 80
```

```python
# Adding 50 to beauty_points
queenB.beauty_plus(50, True)

# Accessing the updated beauty_points
print(queenB.beauty_points)  # Output: 130
```

```python
queenB.beauty_points
```

```python
queenB.beauty_plus(1, False)
```

```python
queenB.beauty_points
```

```python
# Create an instance of Butterfly with name 'Arthur', color 'Blood Red', and season 'Winter'
kingB = Butterfly('Arthur', 'Blood Red', 'Winter')
```

```python
# Accessing the 'beauty_points' attribute of kingB
print(kingB.beauty_points)  # Output: 100
```

```python
# Adding the beauty points of queenB and kingB
total_beauty = queenB + kingB

# Printing the result
print(f"Total beauty points: {total_beauty}")  # Output: 229
```
