### Text / Markdown Cells

In Jupyter Notebooks, we can also create **text cells**. These cells are used for writing notes, explanations, headings, and other content alongside our code.

Python does **not** execute anything inside a text cell. Instead, the text is displayed as formatted content.

Text cells use **Markdown**, a simple syntax for formatting text. They also support **LaTeX**, which is commonly used to write mathematical notation and equations.

For example, we can write a mathematical expression like:

$$
\text{Mean Squared Error}
$$

Or a more complicated equation:

$$
\text{error}(\theta_0,\theta_1)
=

\frac{1}{2m}
\sum_{i=0}^{m-1}
\left(
(h_{\theta_0} + h_{\theta_1}x_1^{(i)}) - y^{(i)}
\right)^2
$$

This means we can keep our **code, explanations, mathematical equations, and even images** together in the same Colab notebook.

That makes notebooks especially useful for learning and experimenting, because we can explain **what our code is doing** right next to the code itself.

We can even add images to our Markdown cells when they help explain an idea.


### Code Cells

Code cells are where we write our python code. Simple enough, right.

**Cell execution**: Press the run/play button located on the leftmost side of the cell


### Keyboard shortcuts

Important **ipynb notebook** (colab) shortcuts



**shift** (hold) + **enter** (press): To execute a cell, hold the shift key on your keyboard and press enter to execute a cell. The cell will be executed and colab will take the user to the cell below. If the next cell doesn't exist, colab will create one.

**ctrl** (hold) + **enter** (press): Execute the current cell and stay on it

**alt** (hold) + **enter** (press): Execute current cell, create a cell below and take the user to the cell below.

### Command mode

If you are writing something inside a colab cell, press **esc** key to get into command mode. From there, you could use following keyboard shortcuts.

* To add cell above: press **a**

* To add cell below: press **b**

* To delete current cell: **d,d** Press d two times in succession

* To undo / retrieve deleted cell: **z,z**

**Specific to ipy kernal of jupyter notebook (not for colab)**

* Change cell type to markdown: **m**

* Change cell type to code: **y**

* Change cell type to raw text: **r**



### Edit mode

If you're typing in a cell, you're in edit mode. If you're in command mode, press enter on a cell to get into the edit more.

Best practices for the variable names

* Variable name must start with a letter

* Variable name should not start with a number

* Variable names should only contain alpha-numerics or underscores (A-Z, a-z, 0-9, _)

```python
# We can choose any variable name but we have to watch out for some certain rules.

# Variable name must start with a letter
variable1 = 10     # is allowed

print(variable1)

# Variable name should not start with a number
# 10variable1 = 11    # is not allowed

# Variable names should only contain alpha-numerics or underscores (A-Z, a-z, 0-9, _)
our_first_variable = 12    # is Allowed
print(our_first_variable) 
```
