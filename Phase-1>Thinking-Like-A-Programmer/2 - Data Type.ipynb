### Escape Sequence

"If I wink, know that I want you to lie". Python understood the words of his creator.


\ backslash character changes the behaviour of whatever character follows the backslash character.

In the following code, 'Alien\'s planet' backslash will tell python to treat the second single quote as a literal quote character rather than using it as a closing quote for the string.

```python
j = 'Alien\'s planet' 
print(j)
```

So, we need to be mindful about using these quotes interchangeably.

We can create strings using either of the two ways, but we need to be careful about how we create these strings. Further discussion on it is kind of, out of the scope for this beginner level tutorial.

```python
#"Alien's planet"  easy and preffered way

#'Alien\'s planet' backslash escape sequence way

#Both works
```

# Multi-Line Strings

There are times where we need to split our strings over multiple lines.

You may find that a lot of people incuding myself use multi-line strings as some sort of comment, as they look more readable than a standard comment.

Much like the one line strings, they also use the quotes, double or single quotes. But we use three start and end quotes for multi-line strings.

```python
"""This is a multi line string we can
    write whatever we want in here 
"""
```

Depending on your system you may notice a \n character, which may dictate the newline character. Python stores the information about new lines by using the \n. 

\n is just one character which means introduce a new line.

```python
multi_line_string = """This is a multi line string we can
    write whatever we want in here 
"""
multi_line_string
```

```python
print(multi_line_string)
```

```python
multi2 = '''This is also a sample
    multi line string but this time starting and ending 
    with three triple quote characters
'''
multi2
```

```python
print(multi2)
```
