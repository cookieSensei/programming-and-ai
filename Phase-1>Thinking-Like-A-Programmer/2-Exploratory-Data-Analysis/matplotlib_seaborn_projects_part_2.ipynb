# Part 1 - Visualization Foundations

Before beginning the projects, establish a mental model for visualization.

## The visual encoding hierarchy

Different visual properties communicate information with different levels of precision.

Generally, viewers are better at comparing:

1. position on a common scale
2. position on different scales
3. length
4. angle
5. area
6. color intensity
7. shape

This is one reason why a carefully designed bar chart is often easier to interpret than a pie chart.

---

## Chart selection cheat sheet

| Analytical question | Recommended chart |
|---|---|
| How does a variable change over time? | Line chart |
| How large is each category? | Bar chart |
| What does a numerical variable look like? | Histogram |
| Are there potential outliers? | Box plot |
| How are two numerical variables related? | Scatter plot |
| How do distributions differ across groups? | Box/violin plot |
| How common is each category? | Count plot |
| How do many numerical variables relate? | Correlation heatmap |
| How does a relationship differ across groups? | Faceted scatter plots |
| What is the distribution shape? | Histogram/KDE |

# Project 9 - Time-Series Visualization

Time is one of the most common dimensions in real-world data.

The business asks:

> **How has monthly revenue changed over time?**

Create a synthetic daily transaction dataset for two years.


````python
dates = pd.date_range(
    "2024-01-01",
    "2025-12-31",
    freq="D"
)

daily = pd.DataFrame({"date": dates})

# Simulate trend + seasonality + noise
day_number = np.arange(len(daily))

trend = 15000 + day_number * 8
seasonality = 2500 * np.sin(2 * np.pi * day_number / 365.25)
weekly = 900 * np.sin(2 * np.pi * day_number / 7)

daily["revenue"] = (
    trend
    + seasonality
    + weekly
    + rng.normal(0, 1200, size=len(daily))
)

daily["revenue"] = daily["revenue"].clip(lower=0)

daily.head()
````

# Project 11 - Correlation Heatmap

Now investigate relationships among multiple numerical variables.

Create a richer customer dataset with:

- age
- income
- spending
- purchase frequency
- average order value
- discount usage
- customer tenure

Then calculate a correlation matrix.


````python
n = 1500

analysis_df = pd.DataFrame({
    "age": rng.integers(18, 70, n),
    "income": rng.lognormal(np.log(70000), 0.5, n),
    "purchase_frequency": rng.poisson(12, n) + 1,
    "avg_order_value": rng.lognormal(np.log(2500), 0.45, n),
    "discount_usage": rng.uniform(0, 1, n),
    "tenure_months": rng.integers(1, 84, n)
})

analysis_df["spending"] = (
    analysis_df["purchase_frequency"]
    * analysis_df["avg_order_value"]
    * rng.normal(1, 0.15, n)
)

analysis_df["spending"] = analysis_df["spending"].clip(lower=0)

analysis_df.head()
````

# What You Should Take Away

Matplotlib and Seaborn are not the skill.

The skill is **visual reasoning**.

You should now be moving from:

> "I know how to make a scatter plot."

to:

> "I know when a scatter plot is useful."

Then from:

> "I found a correlation."

to:

> "I found an association, and I understand what it does and does not establish."

And finally from:

> "Here are five charts."

to:

> **"Here is a question, here is the evidence, here is what the evidence suggests, here are the limitations, and here is what I would investigate next."**

That is the foundation of strong Exploratory Data Analysis.
