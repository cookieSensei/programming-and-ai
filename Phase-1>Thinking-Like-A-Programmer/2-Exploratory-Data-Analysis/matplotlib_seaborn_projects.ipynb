
# Matplotlib and Seaborn Complete Guide

This notebook teaches **data visualization in Python** using:

- Matplotlib
- Seaborn

It also includes **two real-world mini projects**:

1. Housing price visualization (housing.csv)
2. Telecom churn analysis (TelecomCustomerChurn.csv)

Libraries used:
- pandas
- numpy
- matplotlib
- seaborn


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Libraries loaded")
```

## Basic Line Plot

```python

x = np.arange(0,10)
y = x**2

plt.figure()
plt.plot(x,y)
plt.title("Line Plot Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

## Scatter Plot

```python

x = np.arange(0,50)
y = np.arange(50,100)

plt.scatter(x,y)
plt.title("Scatter Plot")
plt.show()
```

## Bar Chart

```python

categories = ["A","B","C"]
values = [10,20,15]

plt.bar(categories,values)
plt.title("Bar Chart")
plt.show()
```

## Histogram

```python

data = np.random.randn(1000)

plt.hist(data,bins=30)
plt.title("Histogram")
plt.show()
```

## Multiple Plots

```python

x = np.linspace(0,10,100)

plt.subplot(1,2,1)
plt.plot(x,np.sin(x))

plt.subplot(1,2,2)
plt.plot(x,np.cos(x))

plt.show()
```

# Seaborn Visualizations

## Distribution Plot

```python

data = np.random.randn(500)

sns.histplot(data,kde=True)
plt.show()
```

## Box Plot

```python

data = pd.DataFrame({
"category":["A","A","B","B","C","C"],
"value":[5,7,3,4,6,8]
})

sns.boxplot(x="category",y="value",data=data)
plt.show()
```

## Heatmap

```python

matrix = np.random.rand(5,5)

sns.heatmap(matrix,annot=True)
plt.show()
```

## Pairplot

```python

df = pd.DataFrame(np.random.rand(100,4),columns=["A","B","C","D"])
sns.pairplot(df)
```


# Project 1: Housing Data Visualization

Dataset: housing.csv

Goals:
- Explore price distribution
- Analyze correlations
- Visualize relationships


```python

housing = pd.read_csv("../datasets/housing.csv")
housing.head()
```

```python

plt.hist(housing.select_dtypes(include=np.number).iloc[:,0], bins=30)
plt.title("Distribution Example")
plt.show()
```

```python

corr = housing.corr(numeric_only=True)

plt.figure(figsize=(10,8))
sns.heatmap(corr,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python

num_cols = housing.select_dtypes(include=np.number).columns[:4]

sns.pairplot(housing[num_cols])
```


# Project 2: Telecom Customer Churn Visualization

Dataset: TelecomCustomerChurn.csv

Goals:
- Understand churn distribution
- Identify patterns in features


```python

churn = pd.read_csv("../datasets/TelecomCustomerChurn.csv")
churn.head()
```

```python

sns.countplot(x="Churn", data=churn)
plt.title("Churn Distribution")
plt.show()
```

```python

numeric_cols = churn.select_dtypes(include=np.number).columns

churn[numeric_cols].hist(figsize=(10,8))
plt.show()
```

```python

plt.figure(figsize=(10,8))
sns.heatmap(churn.corr(numeric_only=True), cmap="viridis")
plt.title("Feature Correlations")
plt.show()
```
