## NumPy Arrays vs Python Lists
Explanation and example.

```python

import numpy as np

python_list=[1,2,3,4]
numpy_array=np.array([1,2,3,4])

print("Python list:",python_list)
print("NumPy array:",numpy_array)

print("List * 2:",python_list*2)
print("Array * 2:",numpy_array*2)
```

## Linear Algebra
Explanation and example.

```python

import numpy as np

A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])

print("Matrix multiplication")
print(A@B)

print("Determinant")
print(np.linalg.det(A))

print("Inverse")
print(np.linalg.inv(A))
```

## Performance: Vectorization vs Loops
Explanation and example.

```python

import numpy as np
import time

x=np.arange(1000000)

start=time.time()
y=[i*i for i in x]
print("Python loop:",time.time()-start)

start=time.time()
y=x*x
print("NumPy vectorized:",time.time()-start)
```
