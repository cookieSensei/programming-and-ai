## GroupBy Example
Explanation and example.

```python

data = {
    "team":["A","A","B","B","B"],
    "points":[10,20,30,40,50]
}

df_group = pd.DataFrame(data)

grouped = df_group.groupby("team").mean()
print(grouped)
```

## Merging DataFrames
Explanation and example.

```python

left = pd.DataFrame({"id":[1,2,3],"name":["Alice","Bob","Charlie"]})
right = pd.DataFrame({"id":[1,2,3],"salary":[50000,60000,70000]})

merged = pd.merge(left,right,on="id")
print(merged)
```

## Pivot Tables
Explanation and example.

```python

data = {
    "city":["NY","NY","London","London"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,130]
}

df_pivot = pd.DataFrame(data)
print(df_pivot)

pivot = df_pivot.pivot_table(values="sales",index="city",columns="year")
print(pivot)
```
